```python
import pandas as pd
import glob
import os

def load_and_average_temperature(file_path_pattern):
    # Get all the CSV files matching the pattern
    all_files = glob.glob(file_path_pattern)
    
    # List to store DataFrames from each file
    data_frames = []
    
    # Loop through each file and read it into a DataFrame
    for file in all_files:
        # Read the CSV file
        dg = pd.read_csv(file, parse_dates=['DATE'])  # Assuming 'Date' column exists
        data_frames.append(dg)
    
    # Concatenate all dataframes row-wise
    combined_df = pd.concat(data_frames)
    
    # Group by 'Date' and calculate the mean for each column
    weekly_average_df = combined_df.groupby('DATE').mean().reset_index()

    # Save the daily averages to a new CSV file
    weekly_average_df.to_csv('daily_average_temperature.csv', index=False)
    
    return weekly_average_df

# Example usage:
# Assuming your CSV files are in the folder named 'data_files/'
file_path_pattern = os.path.join('temperature', '*.csv')

# Load data and calculate daily averages across all files
weekly_average_df = load_and_average_temperature(file_path_pattern)

# Display the resulting daily averages
print(weekly_average_df)
```

```python
# Assuming daily_average_df is the DataFrame you already have

# Save the daily averages to a new Excel file
weekly_average_df.to_excel('weekly_average_temperature.xlsx', index=False)

print("Data saved to 'weekly_average_temperature.xlsx'")
```

```python
import pandas as pd

# Define the function to aggregate data and skip formatting issues
def aggregate_to_weekly(df):
    # Initialize a list to collect valid rows
    valid_rows = []

    # Iterate over each row to handle date and time formatting
    for index, row in df.iterrows():
        try:
            # Combine the Date and time into a DateTime column
            row['DateTime'] = pd.to_datetime(row['Date'] + ' ' + row['time'], format="%m/%d/%Y %H:%M")
            valid_rows.append(row)  # If successful, append the row to valid_rows
        except ValueError:
            # Skip any row with a formatting issue (do nothing)
            continue

    # Create a new DataFrame from valid rows
    valid_df = pd.DataFrame(valid_rows)

    # Drop unnecessary columns ('status' and 'unit')
    columns_to_keep = ['DateTime', 'Carbon monoxide', 'Nitric oxide', 'PM<sub>10</sub> particulate matter (Hourly measured)',
                       'Nitrogen dioxide', 'Nitrogen oxides as nitrogen dioxide', 'Ozone', 
                       'PM<sub>2.5</sub> particulate matter (Hourly measured)', 'Sulphur dioxide']
    valid_df = valid_df[columns_to_keep]

    # Set the DateTime as index
    valid_df.set_index('DateTime', inplace=True)

    # Resample the data to daily averages (from hourly to daily)
    daily_avg = valid_df.resample('D').mean()

    # Resample again to weekly averages and ensure weeks start on Thursday
    weekly_avg = daily_avg.resample('W-THU').mean()

    # Return the final weekly-aggregated DataFrame
    return weekly_avg

# Step 1: Read the CSV file into a DataFrame
air_data = pd.read_csv('MY1_2024.csv')

# Step 2: Call the function to aggregate data into weekly averages
weekly_avg_data = aggregate_to_weekly(air_data)

# Step 3: Save the weekly average data to a new CSV file
weekly_avg_data.to_csv('weekly_avg_air.csv')

print("Weekly average data saved as 'weekly_avg_air.csv'")
```

```python
import pandas as pd

# Define the function to aggregate data and handle date-time formatting issues
def aggregate_to_weekly(df):
    # Try to convert the 'Date' and 'time' columns into a DateTime column, but ignore errors
    df['DateTime'] = pd.to_datetime(df['Date'] + ' ' + df['time'], format="%m/%d/%Y %H:%M", errors='coerce')

    # Drop unnecessary columns ('status' and 'unit')
    columns_to_keep = ['DateTime', 'Carbon monoxide', 'Nitric oxide', 'PM<sub>10</sub> particulate matter (Hourly measured)',
                       'Nitrogen dioxide', 'Nitrogen oxides as nitrogen dioxide', 'Ozone', 
                       'PM<sub>2.5</sub> particulate matter (Hourly measured)', 'Sulphur dioxide']
    df = df[columns_to_keep]

    # Drop rows where DateTime conversion failed (NaT values)
    valid_df = df.dropna(subset=['DateTime'])

    # Set the DateTime as index
    valid_df.set_index('DateTime', inplace=True)

    # Resample the data to daily averages (from hourly to daily)
    daily_avg = valid_df.resample('D').mean()

    # Resample again to weekly averages and ensure weeks start on Thursday
    weekly_avg = daily_avg.resample('W-THU').mean()

    # Return the final weekly-aggregated DataFrame
    return weekly_avg

# Step 1: Read the CSV file into a DataFrame
air_data = pd.read_csv('MY1_2024.csv')

# Step 2: Call the function to aggregate data into weekly averages
weekly_avg_data = aggregate_to_weekly(air_data)

# Step 3: Save the weekly average data to a new CSV file
weekly_avg_data.to_csv('weekly_avg_air1.csv')

print("Weekly average data saved as 'weekly_avg_air1.csv'")
```

```python
import pandas as pd

# Define the function to aggregate data and handle partial missing data
def aggregate_to_weekly(df):
    # Try to convert the 'Date' and 'time' columns into a DateTime column, skipping any issues
    df['DateTime'] = pd.to_datetime(df['Date'] + ' ' + df['time'], format="%m/%d/%Y %H:%M", errors='coerce')

    # Drop unnecessary columns ('status' and 'unit')
    columns_to_keep = ['DateTime', 'Carbon monoxide', 'Nitric oxide', 'PM<sub>10</sub> particulate matter (Hourly measured)',
                       'Nitrogen dioxide', 'Nitrogen oxides as nitrogen dioxide', 'Ozone', 
                       'PM<sub>2.5</sub> particulate matter (Hourly measured)', 'Sulphur dioxide']
    df = df[columns_to_keep]

    # Set the DateTime as index, but keep the invalid DateTime rows for further calculations
    valid_df = df.dropna(subset=['DateTime'])

    # Resample to calculate the mean for each day (it will use the available hours)
    daily_avg = valid_df.resample('D', on='DateTime').mean()

    # Resample again to weekly averages and ensure weeks start on Thursday
    weekly_avg = daily_avg.resample('W-THU').mean()

    # Restrict the output to the date range of the original data (ending at the latest valid date)
    last_valid_date = valid_df['DateTime'].max()
    weekly_avg = weekly_avg[weekly_avg.index <= last_valid_date]

    # Return the final weekly-aggregated DataFrame
    return weekly_avg


```

```python
# Step 1: Read the CSV file into a DataFrame
air_data = pd.read_csv('MY1_2024.csv')

# Step 2: Call the function to aggregate data into weekly averages
weekly_avg_data = aggregate_to_weekly(air_data)

# Step 3: Save the weekly average data to a new CSV file
weekly_avg_data.to_csv('weekly_avg_air.csv')

print("Weekly average data saved as 'weekly_avg_air.csv'")
```
